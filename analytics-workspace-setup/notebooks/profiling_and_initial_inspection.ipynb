# SellerSignal Dataset Cleaning & Analysis

This notebook contains the step-by-step cleaning, profiling, and outlier handling process for the messy `SellerSignal` raw dataset.

## Step 1: Profiling and Initial Inspection
Let's load the four raw CSV datasets and print their shapes, column info, missing value counts, and duplicate counts.

```python
import pandas as pd
import numpy as np

# Load datasets
sellers = pd.read_csv('raw_sellers.csv')
orders = pd.read_csv('raw_orders.csv')
returns = pd.read_csv('raw_returns.csv')
reviews = pd.read_csv('raw_reviews.csv')

# Print shapes
print("sellers shape:", sellers.shape)
print("orders shape:", orders.shape)
print("returns shape:", returns.shape)
print("reviews shape:", reviews.shape)
```

```python
print("=== Sellers Info ===")
sellers.info()
print("\nMissing values:")
print(sellers.isnull().sum())
print("\nDuplicate rows count:", sellers.duplicated().sum())

print("\n=== Orders Info ===")
orders.info()
print("\nMissing values:")
print(orders.isnull().sum())
print("\nDuplicate rows count:", orders.duplicated().sum())

print("\n=== Returns Info ===")
returns.info()
print("\nMissing values:")
print(returns.isnull().sum())
print("\nDuplicate rows count:", returns.duplicated().sum())

print("\n=== Reviews Info ===")
reviews.info()
print("\nMissing values:")
print(reviews.isnull().sum())
print("\nDuplicate rows count:", reviews.duplicated().sum())
```
