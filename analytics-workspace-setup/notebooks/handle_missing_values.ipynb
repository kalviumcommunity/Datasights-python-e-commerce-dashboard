## Step 5: Handle Missing Values
We will impute missing values using simple placeholders to keep the dataset consistent and clean.

```python
# Fill missing categorical/text values with standard placeholders
sellers['region'] = sellers['region'].fillna('Unknown')
orders['customer_id'] = orders['customer_id'].fillna('Unknown')
returns['return_reason'] = returns['return_reason'].fillna('Unknown')
reviews['review_text'] = reviews['review_text'].fillna('No Review Text')

# Print remaining null counts to verify
print("Sellers null counts:")
print(sellers.isnull().sum())
print("\nOrders null counts:")
print(orders.isnull().sum())
print("\nReturns null counts:")
print(returns.isnull().sum())
print("\nReviews null counts:")
print(reviews.isnull().sum())
```
